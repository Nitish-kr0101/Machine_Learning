```python
import numpy as np
import pandas as pd
 
df=pd.read_csv('employees.csv')
print(df)
```

```python
print(df["salary"])
```

```python
df.info()
```

```python
print(df[["salary","designation"]])
```

```python
#print salary more than 50000
print(df[df["salary"]>50000])
#count of employees drawing salary more than 50000 
print(df[df["salary"]>50000].shape[0])
#print it employee drawing salary more than 50000 and designation as it
print(df[(df["salary"]>50000) & (df["designation"]=="it")])
```

```python
#print withdrawing max salary
print(df[df["salary"]==df["salary"].max()])
```

```python
#display employee in salary descending order
print(df.sort_values(by="salary",ascending=False))
```

```python
#in each city how many employees are there
print(df["city"].value_counts())
```

```python
#row in which salary is missing
print(df[df["salary"].isnull()])
#fill missing salary with average salary
df["salary"].fillna(df["salary"].mean(),inplace=True)
#push the changes to csv
df.to_csv('employees.csv',index=False)
```

```python
#add new column bonus commmon for all employees
df["bonus"]=5000
#drop bonus column
df.drop("bonus",axis=1,inplace=True)
print(df)
#give hike of 10% for it and 5% for hr department and give new salary column name as new_salary
df["new_salary"]=np.where(df["designation"]=="it",df["salary"]*1.1,np.where(df["designation"]=="hr",df["salary"]*1.05,df["salary"]))
print(df)
```

```python
df["department"] = df["department"].replace("IT", 1)
df["department"] = df["department"].replace("Finance", 2)
df["department"] = df["department"].replace("HR", 3)
df["department"] = df["department"].replace("Sales", 4)
df["department"] = df["department"].replace("Operations", 5)


df1=pd.read_csv('department.csv')
print(df1)
 
df=df.merge(df1, on='department', how='inner')
print(df)
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['experience'],df['salary'])
plt.xlabel('Experience')
plt.ylabel('Salary')
plt.title('Experience vs Salary')
plt.show()
```

```python
#histogram for employee of different age groups
plt.hist(df['age'], bins=[20,25,30,35,40,45,50,55,60])
```

```python
#for each city how many employees are there
plt.bar(df['city'].value_counts().index,df['city'].value_counts().values)
plt.xlabel('City')
```

```python
#usng seaborn to plot
import seaborn as sns
sns.boxplot(x='designation',y='salary',data=df)
plt.title('Designation vs Salary')
plt.show()
sns.boxplot(x='city',y='salary',data=df)
plt.title('City vs Salary')
plt.show()
sns.boxplot(x='department',y='salary',data=df)
plt.title('Department vs Salary')
plt.show() 
sns.histplot(df['age'], bins=[20,25,30,35,40,45,50,55,60], kde=True)
plt.title('Age Distribution')   
```
