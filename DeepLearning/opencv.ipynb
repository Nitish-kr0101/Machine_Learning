```python

import cv2
import matplotlib.pyplot as plt

image = cv2.imread("image.jpg")   

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

blur = cv2.GaussianBlur(gray, (5, 5), 0)

edges = cv2.Canny(blur, 100, 200)



plt.figure(figsize=(12,6))

plt.subplot(1,4,1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1,4,2)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale")
plt.axis("off")

plt.subplot(1,4,3)
plt.imshow(blur, cmap='gray')
plt.title("Blur")
plt.axis("off")

plt.subplot(1,4,4)
plt.imshow(edges, cmap='gray')
plt.title("Edges")
plt.axis("off")

plt.show()
```

```python
# read content from id_sample.jpg using pytesseract
import pytesseract
from PIL import Image
import os
path = 'id_sample.jpg'
if not os.path.exists(path):
    print(f"{path} not found in notebook directory.")
else:
    img = cv2.imread(path)
    if img is None:
        print('Failed to read image (cv2.imread returned None). Check file format or permissions.')
    else:
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        pil_img = Image.fromarray(img_rgb)
        try:
            text = pytesseract.image_to_string(pil_img)
            print(text)
        except Exception as e:
            print('pytesseract error:', e)
```
